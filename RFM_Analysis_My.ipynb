```python
import pandas as pd

# 假設您的原始 DataFrame 長這樣
# df = pd.DataFrame({
#     'CustomerKey': ['C001', 'C001', 'C002', 'C003', 'C003'],
#     'Date': ['2026-05-01', '2026-06-01', '2026-05-15', '2026-04-01', '2026-06-10'],
#     'InvoNo': ['INV001', 'INV002', 'INV003', 'INV004', 'INV005'],
#     'Amount': [1000, 1500, 2000, 500, 1200]
# })

# 1. 確保「Date」欄位是 datetime 型態，以便後續做Date減法
df['Date'] = pd.to_datetime(df['Date'])

# 2. 設定基準日（R 的計算基準）。
# 這裡以資料集中的「最大Date」作為基準日（您也可以根據需求換成今天 pd.Timestamp.today()）
max_date = df['Date'].max()

# 3. 使用 groupby 進行聚合運算
rfm_df = df.groupby('CustomerKey').agg(
    Customer=('CustomerKey', 'first'),                     # 保留CustomerKey
    F=('InvoNo', 'count'),                        # InvoNo計數 (Frequency)
    M=('Amount', 'sum'),                              # Amount加總 (Monetary)
    R_date=('Date', 'max')                          # 記錄該客戶最近一次交易Date
).reset_index(drop=True)

# 4. 計算 R：基準日與該客戶最大交易Date的差值（以天數表示）
rfm_df['R'] = (max_date - rfm_df['R_date']).dt.days

# 5. 如果不需要中間產生的 'R_date' 欄位，可以把它drop掉
rfm_df = rfm_df[['Customer', 'F', 'M', 'R']]

# 檢視結果
print(rfm_df)
```
